## Data Analysis

```python
import pandas as pd
df = pd.read_csv('Coffee_company.csv')
df.head()
```

```python
# clean up data
# colnames
df.columns = df.columns.str.strip()
```

```python
print(df.info())
```

```python
# convert columns to numeric
numeric_cols = ['Manufacturing Price', 'Sale Price', 
                'Gross Sales', 'Discounts',
                'Sales', 'COGS', 'Profit']

df[numeric_cols]
```

In this case, I need to remove the dollar sign, negatives

```python
df1 = df.copy()
df1[numeric_cols]

for col in numeric_cols:
    df1[col] = df1[col].str.strip().str.replace("\$|,|-|\(|\)", '', regex=True)
    df1[col] = df1[col].replace('', float('nan'))
    df1[col] = df1[col].apply(lambda x: float(x))
    print(col, '...done')
```

```python
# fix date
df1['Date'] = pd.to_datetime(df1['Date'], format='%d/%m/%Y')
```

```python
# total sales
print('Total sales ->', round(df1['Sales'].sum(), 2))
print('Average sales ->', df1['Sales'].mean())

# quarterly sales
df1.set_index('Date', inplace=True)

# Calculate quarterly sales
quarterly_sales = df1['Sales'].resample('QE').sum().reset_index()
quarterly_sales.columns = ['Quarter', 'Total Sales']

# restore index
df1.reset_index(inplace=True)

print('quarterly sales ...',quarterly_sales)
```

```python
# analytics
max_sales = quarterly_sales['Total Sales'].max()
print('Highest sales of', 
      max_sales,
      'in period',quarterly_sales[quarterly_sales['Total Sales']==max_sales]['Quarter'].dt.to_period('Q').astype(str))

min_sales = quarterly_sales['Total Sales'].min()
print('Lowest sales of',
      min_sales,
      'in period',quarterly_sales[quarterly_sales['Total Sales']==min_sales]['Quarter'].dt.to_period('Q').astype(str))
```

```python
# plot
plt.bar(quarterly_sales['Quarter'].dt.to_period('Q').astype(str), quarterly_sales['Total Sales'], color='skyblue')

# Set the properties of the plot
plt.title('Quarterly Sales')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()

# show the plot (you need this for VS Code)
plt.show()

# highest sales are seen in 2023Q4, lowest in 
# average sales are $259,057.

# Adding data labels on top of the bars
bars  = plt.bar(quarterly_sales['Quarter'].dt.to_period('Q').astype(str), quarterly_sales['Total Sales'], color='skyblue')
# loop through each bar to add the label in the specified location
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, yval, round(yval, 2), ha='center', va='bottom')
plt.show()
```

```python
# plot
plt.bar(segment_sales['Segment'], height=segment_sales['Sales'])
plt.xticks(rotation=45)
plt.title('Sales by segment')
```

# Forecasting

```python
# Time series forecasting without a dataframe
# %%
# Question 5 - Linear Regression for Forecasting
# Using the dataset below, create a linear regression model to predict the demand for given prices.
# Predict the demand if the company sets the price at £26. Show a scatter plot of the data points and plot the regression line.
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Price (£): 15, 18, 20, 22, 25, 27, 30
# Demand (Units): 200, 180, 170, 160, 150, 140, 130
prices = np.array([15, 18, 20, 22, 25, 27, 30]).reshape(-1, 1)
demand =  np.array([200, 180, 170, 160, 150, 140, 130])

model = LinearRegression()
model.fit(prices, demand)

predicted_demand = model.predict(np.array([[26]]))

price_range = np.linspace(15, 30, 100).reshape(-1, 1)
demand_predictions = model.predict(price_range)

# Plot scatter and regression line
plt.scatter(prices, demand, color='blue', label='Actual Data')
plt.plot(price_range, demand_predictions, color='red', label='Regression Line')
plt.scatter([26], predicted_demand, color='green', marker='x', s=100, label=f'Predicted Demand ({predicted_demand[0]:.2f})')
plt.xlabel("Price (£)")
plt.ylabel("Demand (Units)")
plt.title("Price vs Demand Linear Regression")
plt.legend()
plt.grid()
plt.show()
```

## Exponential Smoothing

| Method                                | Components Used         | When to Use?                  | Python Example |
|--------------------------------------|------------------------|-------------------------------|---------------|
| **Simple Exponential Smoothing (SES)** | Level                  | No trend or seasonality       | `fit = SimpleExpSmoothing(series).fit(smoothing_level=0.2)` |
| **Holt’s Method (Double Smoothing)**  | Level + Trend          | Trend but no seasonality      | `fit = ExponentialSmoothing(series, trend="add").fit(smoothing_level=0.2, smoothing_slope=0.1)` |
| **Holt-Winters (Triple Smoothing)**   | Level + Trend + Seasonality | Both trend and seasonality | `fit = ExponentialSmoothing(series, trend="add", seasonal="multiplicative", seasonal_periods=12).fit(smoothing_level=0.2, smoothing_slope=0.1, smoothing_seasonal=0.05)` |


```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
import matplotlib.pyplot as plt
```

```python
monthly_sales = df1.groupby('Date')['Sales'].sum().reset_index()
monthly_sales.head()
```

```python
sns.lineplot(data=monthly_sales, x='Date', y='Sales')
```

```python
model = ExponentialSmoothing(
    monthly_sales['Sales'],
    trend='additive',
    seasonal='multiplicative',
    seasonal_periods=12
)
fit = model.fit()
```

```python
# forecast
# Step 3: Forecast for the next 12 months
forecast_periods = 24
forecast = fit.forecast(forecast_periods)

# Step 4: Prepare DataFrame for Forecasted Periods
last_date = monthly_sales['Date'].iloc[-1]
forecast_dates = pd.date_range(last_date + pd.offsets.MonthEnd(1), periods=forecast_periods, freq='M')
forecast_df = pd.DataFrame({'Date': forecast_dates, 'Sales': forecast})

# Step 5: Combine Actual and Forecasted Data
all_data = pd.concat([
    monthly_sales,
    forecast_df
]).reset_index(drop=True)
```

```python
# Step 6: Plot Actual and Forecasted Sales
# Note that these are two plots plotted on the same figure
plt.figure(figsize=(14, 7))
plt.plot(monthly_sales['Date'], monthly_sales['Sales'], label='Actual Monthly Sales', marker='o', color='b')
plt.plot(forecast_df['Date'], forecast_df['Sales'], label='Forecasted Monthly Sales', marker='o', linestyle='--', color='r')
plt.axvline(x=last_date, color='gray', linestyle='--', label='Forecast Start')
plt.title('Actual and Forecasted Monthly Sales (Next 12 Months)')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.legend()
plt.grid(True)
plt.show()
```
